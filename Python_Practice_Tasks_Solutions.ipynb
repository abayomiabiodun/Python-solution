# Part A: Variables, Data Types & Operators

## Task 1: Mama Ngozi's Provision Store

**Explanation:** We store the three inputs with the data types the task specifies (`str`, `int`, `float`), then compute `total_cost = price_per_item * items_bought`. We use a comparison operator (`>`) to decide whether the ₦5,000 threshold is crossed, and if so apply a 5% discount (`total_cost * 0.05`). The final amount is `total_cost - discount`. An f-string formats the receipt, with `:.2f` used to keep money values to 2 decimal places.

```python
customer_name = "Chioma"          # str
items_bought = 12                 # int
price_per_item = 550.0            # float (Naira)

total_cost = price_per_item * items_bought

# Comparison operator to check the discount threshold
if total_cost > 5000:
    discount = total_cost * 0.05  # 5% discount
else:
    discount = 0.0

final_amount = total_cost - discount

print("----- MAMA NGOZI'S PROVISION STORE -----")
print(f"Customer: {customer_name}")
print(f"Items bought: {items_bought}")
print(f"Total cost: ₦{total_cost:.2f}")
print(f"Discount applied: ₦{discount:.2f}")
print(f"Final amount to pay: ₦{final_amount:.2f}")
```

## Task 2: The School Registration Form

String **concatenation** (with `+`) needs all parts to be strings, so `str(age)` isn't needed here since we only concatenate the name into a welcome message. Finally, an f-string neatly prints everything together.

## Task 3: Danfo Bus Fare Calculator

**Explanation:** `total_fare = fare_per_passenger * num_passengers` uses the `*` operator. The `%` (modulus) operator gives the remainder when dividing by 100 — if that remainder is `0`, the fare is an exact multiple of ₦100, meaning exact change (in ₦100 notes) is possible.

```python
fare_per_passenger = 250.0   # float
num_passengers = 14          # int

total_fare = fare_per_passenger * num_passengers

# Modulus operator checks divisibility by 100
remainder = total_fare % 100

print(f"Total fare: ₦{total_fare:.2f}")
if remainder == 0:
    print("Exact change (multiples of ₦100) is possible.")
else:
    print(f"Exact change is NOT possible. Remainder of ₦{remainder:.2f} left over.")
```

## Task 4: Football Match Scoreboard

**Explanation:** We compare `team_a_score` and `team_b_score` using `>`, `<`, and `==` to decide the outcome, and compute the goal difference with subtraction. `abs()` keeps the difference positive regardless of which team we subtract from which.

```python
team_a = "Rangers FC"   # str
team_b = "Eagles FC"    # str
team_a_score = 3        # int
team_b_score = 1        # int

goal_difference = abs(team_a_score - team_b_score)

if team_a_score > team_b_score:
    print(f"Winner: {team_a} | Loser: {team_b} | Goal difference: {goal_difference}")
elif team_b_score > team_a_score:
    print(f"Winner: {team_b} | Loser: {team_a} | Goal difference: {goal_difference}")
else:
    print(f"It's a draw between {team_a} and {team_b}! Goal difference: {goal_difference}")
```

## Task 5: Simple Interest Savings Plan

**Explanation:** Simple interest is calculated with the standard formula `(Principal * Rate * Time) / 100`. The total amount is principal plus the interest earned. `:.2f` in the f-string rounds both values to 2 decimal places for currency display.

```python
principal = 150000.0   # float (Naira)
rate = 8.5              # float (percent)
time_years = 3           # int

interest = (principal * rate * time_years) / 100
total_amount = principal + interest

print(f"Interest earned: ₦{interest:.2f}")
print(f"Total amount after {time_years} years: ₦{total_amount:.2f}")
```

## Task 6: Exam Pass or Fail Checker

**Explanation:** `>=` checks the pass mark of 50, and the logical operator `and` additionally ensures the score doesn't exceed 100 (a sanity check for a valid score). Only when both conditions are true is the score treated as a genuine pass.

```python
student_name = "Ifeoma"   # str
score = 68                 # int (out of 100)

# Comparison + logical operator (and) to validate and check for a pass
if score >= 50 and score <= 100:
    print(f"{student_name} PASSED with a score of {score}.")
elif score > 100:
    print(f"Invalid score entered for {student_name}: {score}.")
else:
    print(f"{student_name} FAILED with a score of {score}.")
```

## Task 7: Airtime Recharge App

**Explanation:** We check with `>=` whether this is the customer's 5th (or later) recharge this month. If so, a 10% bonus (`recharge_amount * 0.10`) is added using multiplication and addition. The total credited is the recharge amount plus any bonus.

```python
phone_number = "0803XXXXXXX"   # str
recharge_amount = 1000.0        # float (Naira)
recharge_count_this_month = 5    # int

if recharge_count_this_month >= 5:
    bonus = recharge_amount * 0.10   # 10% bonus airtime
else:
    bonus = 0.0

total_credited = recharge_amount + bonus

print(f"Phone number: {phone_number}")
print(f"Recharge amount: ₦{recharge_amount:.2f}")
print(f"Loyalty bonus: ₦{bonus:.2f}")
print(f"Total credited: ₦{total_credited:.2f}")
```

## Task 8: Recipe Ingredient Scaler

**Explanation:** To scale a recipe, we find the rice needed *per guest* (`cups_of_rice / original_servings`) then multiply by the new number of guests — this is done in one division-based expression: `(cups_of_rice / original_servings) * new_guests`.

```python
original_servings = 4     # int
cups_of_rice = 3.0         # float
new_guests = 10             # int

rice_per_guest = cups_of_rice / original_servings
new_rice_needed = rice_per_guest * new_guests

print(f"For {new_guests} guests, you need {new_rice_needed:.2f} cups of rice.")
```

## Task 9: Electricity Bill Estimator

**Explanation:** The base bill is `units_consumed * cost_per_unit`. We use `>` to compare against the ₦10,000 threshold; if exceeded, a 2% surcharge (`bill * 0.02`) is calculated with `*` and added with `+` to get the final bill.

```python
customer_name = "Mr. Okafor"   # str
units_consumed = 320.0          # float (kWh)
cost_per_unit = 45.0             # float (Naira)

bill = units_consumed * cost_per_unit

if bill > 10000:
    surcharge = bill * 0.02   # 2% surcharge
    final_bill = bill + surcharge
else:
    surcharge = 0.0
    final_bill = bill

print(f"Customer: {customer_name}")
print(f"Base bill: ₦{bill:.2f}")
print(f"Surcharge: ₦{surcharge:.2f}")
print(f"Final bill amount: ₦{final_bill:.2f}")
```

## Task 10: Age in Days Calculator

**Explanation:** Days lived is approximated with `age_years * 365`. Floor division `//` on `age_years // 10` gives the number of complete decades lived, discarding any remainder (e.g. 27 // 10 = 2 complete decades).

```python
name = "Chidi"     # str
age_years = 12       # int

days_lived = age_years * 365
decades_lived = age_years // 10   # floor division

print(f"Hey {name}! You are approximately {days_lived} days old, "
      f"and you've lived through {decades_lived} complete decade(s). 🎉")
```

# Part B: Data Structures — Student Course Registration

We use the four core Python collection types, each suited to a different job:

| Structure | Ordered? | Mutable? | Duplicates? | Access by |
|---|---|---|---|---|
| List `[]` | Yes | Yes | Yes | index |
| Tuple `()` | Yes | No | Yes | index |
| Set `{}` | No | Yes (but elements must be unique/hashable) | No | membership only |
| Dictionary `{k: v}` | Yes (insertion order, Python 3.7+) | Yes | Keys unique | key |


## Part 1: Lists `[]` — Ordered, Mutable

## Part 2: Tuples `()` — Ordered, Immutable

```python
# 5. Create a tuple called student_id
student_id = (2026, "OAU", 15)
print(student_id)
```

**6. Try to change the last element to 16. What happens?**

Explanation: Tuples are **immutable** — once created, their contents cannot be changed, added to, or removed. Attempting `student_id[2] = 16` raises a `TypeError: 'tuple' object does not support item assignment`, because tuples don't implement item assignment the way lists do.

```python
# Demonstrating the error (caught so the notebook keeps running)
try:
    student_id[2] = 16
except TypeError as e:
    print(f"Error: {e}")
```

```python
# 7. Print how many items are in the tuple
print(len(student_id))
```

**10. Why does "python" only appear once?**

Explanation: A set automatically discards duplicate values because, by definition, sets only store **unique** elements. When `"python"` was listed twice in the original set literal, Python silently kept only one copy — sets use hashing to check for and eliminate duplicates.

```python
print(skills)
print(f"Number of unique skills: {len(skills)}")
```

## Part 5: Quick Thinking

**15. Which data structure would you use to store:**

- **a. A list of unique emails collected from a form?**
  A **set** — since emails collected from a form may contain accidental duplicates, and a set automatically enforces uniqueness, guaranteeing each email appears only once.

- **b. Coordinates of a location that should never change?**
  A **tuple** — coordinates like `(latitude, longitude)` represent a fixed pair that shouldn't be accidentally modified, and a tuple's immutability protects that data from being changed after creation.

- **c. Student grades where you need to look up by subject name?**
  A **dictionary** — because you need to retrieve a grade *by subject name* (a key) rather than by position, a dictionary's key-value mapping (e.g. `{"Python": "A", "Math": "B"}`) gives direct, fast lookup by name.


```python
# Illustrating the answers with real code

# a. Unique emails -> set
emails = {"tolu@example.com", "ada@example.com", "tolu@example.com"}
print("Unique emails:", emails)

# b. Fixed coordinates -> tuple
location = (6.5244, 3.3792)   # e.g. Lagos, Nigeria
print("Coordinates (immutable):", location)

# c. Grades by subject -> dictionary
grades = {"Python": "A", "Math": "B", "Statistics": "A"}
print("Grade for Statistics:", grades["Statistics"])
```
